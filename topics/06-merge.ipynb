# merge(), join() y concat()

En Pandas, combinar DataFrames es una operación fundamental para trabajar con información almacenada en diferentes tablas como clientes, ventas, productos o empleados

- merge(): Sirve para combinar DF utilizando columnas en comun 
- join(): Combinar DF mediante sus indices
- concat(): Unir DF vertical u horizontalmente

---

## Crear DF

```python
import pandas as pd

clientes = pd.DataFrame({
    "id_cliente": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Marta", "Sofia"],
    "ciudad": ["Aguascalientes", "Leon", "Zacatecas", "Aguascalientes"]
})

ventas = pd.DataFrame({
    "id_venta": [101, 102, 103, 104],
    "id_cliente": [1, 2, 1, 5],
    "monto": [1500, 500, 800, 1200]
})
```

### Inner

Conserva filas cuya clave existe en ambos df

```python
resultado = pd.merge(
    clientes,
    ventas,
    on="id_cliente",
    how="inner"
)

resultado
```

### left
Conserva todo lo de la izquierda independiente si tiene lo de la derecha

```python
resultado = pd.merge(
    clientes,
    ventas,
    on="id_cliente",
    how="left"
)

resultado
```

### right

Conserva todo derecha independientemente de izquierda

```python
resultado = pd.merge(
    clientes,
    ventas,
    on="id_cliente",
    how="right"
)

resultado
```

### Outer
Incluye todo

```python
resultado = pd.merge(
    clientes,
    ventas,
    on="id_cliente",
    how="outer"
)

resultado
```

---

## join()

Alternativa a merge() cuando los datos están relacionados por sus indices

```python
clientes_idx = clientes.set_index("id_cliente")
ventas_idx = ventas.set_index("id_cliente")

resultado = clientes_idx.join(
    ventas_idx,
    how="left"
)

resultado
```

## pd.concat()

Simplemente concatena DF a lo largo de un eje

### Unir verticalmente (axis = 0)

```python
enero = pd.DataFrame({
    "producto": ["Laptop", "Mouse"],
    "ventas": [15000, 500]
})

febrero = pd.DataFrame({
    "producto": ["Monitor", "Teclado"],
    "ventas": [4000, 800]
})

resultado = pd.concat(
    [enero, febrero],
    ignore_index=True
)

resultado
```

### Unir horizontalmente (axis = 1)

```python
datos = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta"]
})

calificaciones = pd.DataFrame({
    "calificacion": [95, 88, 92]
})

resultado = pd.concat(
    [datos, calificaciones],
    axis=1
)

resultado
```
