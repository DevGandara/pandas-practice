## Crear DF

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "id_venta": [101, 102, 103, 104, 105, 106,
                 107, 108, 109, 110, 111, 112],
    "fecha": [
        "2026-01-03", "2026-01-05", "2026-01-06",
        "2026-01-12", "2026-01-19", "2026-01-20",
        "2026-02-01", "2026-02-04", "2026-02-09",
        "2026-02-12", "2026-02-15", "2026-02-20"
    ],
    "sucursal": [
        "Norte", "Sur", "Norte", "Centro",
        "Sur", "Centro", "Norte", "Sur",
        "Centro", "Norte", "Sur", "Centro"
    ],
    "producto": [
        "Laptop", "Mouse", "Teclado", "Monitor",
        "Laptop", "Mouse", "Laptop", "Monitor",
        "Teclado", "Mouse", "Laptop", "Monitor"
    ],
    "precio": [
        15000, 350, 850, 4500, 14500, 320,
        15800, 4800, 900, 300, 14900, 4400
    ],
    "unidades": [1, 4, 2, 1, 2, 5, 1, 1, 3, 4, 1, 2],
    "satisfaccion": [
        5, 4, 4, 5, 5, 3,
        5, np.nan, 4, 3, 5, 4
    ]
}

df = pd.DataFrame(data)

# Agregamos un duplicado intencional para practicar
df = pd.concat([df, df.iloc[[-1]]], ignore_index=True)

df
```

### Estadisticas individuales

```python
df["precio"].mean()       # Promedio
df["precio"].median()     # Mediana
df["precio"].mode()       # Moda
df["precio"].std()        # Desviación estándar
df["precio"].min()        # Mínimo
df["precio"].max()        # Máximo
df["precio"].quantile(.25) # Primer cuartil
```

## Pivot table
Tablas dinamicas de Excel

```python
df["total"] = df["precio"] * df["unidades"]

tabla = pd.pivot_table(
    df,
    values="total",
    index="sucursal",
    columns="producto",
    aggfunc="sum",
    fill_value=0
)

tabla
```

## corr()
Mide la intensidad y dirección de la relación lineal entre dos variables numéricas

Usa Pearson:
- 1: positiva fuerte
- -1: negativa fuerte
- 0: debil o inexistente

```python
df[["precio", "unidades", "total"]].corr()
```

## nlargest() / nsmallest()

```python
# Las 3 ventas con mayores ingresos
df.nlargest(3, "total")

# Las 3 ventas con menores ingresos
df.nsmallest(3, "total")
```
