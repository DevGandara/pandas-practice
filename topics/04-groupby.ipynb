## Group by

Se utiliza para agrupar filas que tienen valores en común y realiza operaciones sobre cada grupo, como sumas, promedios, conteos o estadisticas
1. Divide los datos en grupos
2. Aplica una operación a cada grupo
3. Combina los resultados en una nueva estructura

```python
import pandas as pd

data = {
    "vendedor": ["Ana", "Luis", "Marta", "Pedro", "Sofia",
                 "Diego", "Laura", "Carlos", "Elena", "Ivan"],
    "sucursal": ["Norte", "Norte", "Sur", "Sur", "Norte",
                 "Sur", "Norte", "Sur", "Norte", "Sur"],
    "producto": ["Laptop", "Mouse", "Laptop", "Mouse", "Laptop",
                 "Laptop", "Mouse", "Mouse", "Laptop", "Mouse"],
    "ventas": [15000, 1000, 14000, 1200, 16000,
               15500, 900, 1100, 17000, 1300],
    "unidades": [3, 10, 2, 12, 3, 3, 9, 11, 4, 13]
}

df = pd.DataFrame(data)
df
```

### Agrupación 1 columna

```python
df.groupby("sucursal")["ventas"].sum()
```

### Agrupación mas columnas

```python
df.groupby("sucursal")[["ventas", "unidades"]].sum()
```

## Agrupación multiples columnas

```python
df.groupby(["sucursal", "producto"])["ventas"].sum()

df.groupby(
    ["sucursal", "producto"],
    as_index=False
)["ventas"].sum() # as_index conserva las columnas utilizadas para agrupar como columnas normales
```

---

## Funciones de agregación mas utilizadas

```python
# Promedio de ventas por sucursal
df.groupby("sucursal")["ventas"].mean()

# Venta máxima por sucursal
df.groupby("sucursal")["ventas"].max()

# Venta mínima por sucursal
df.groupby("sucursal")["ventas"].min()

# Número de registros por sucursal
df.groupby("sucursal").size()

# Cantidad de productos diferentes por sucursal
df.groupby("sucursal")["producto"].nunique()

df["producto"].value_counts()
# sum(), count(), median(), std()
```

---

## agg()

Permite aplicar diferentes operaciones en una sola agrupación

```python
df.groupby("sucursal")["ventas"].agg([
    "sum",
    "mean",
    "max",
    "min",
    "count"
])
```

Tambien se puede aplicar operaciones distintas a diferentes columnas

```python
df.groupby("sucursal").agg({
    "ventas": ["sum", "mean", "max"],
    "unidades": ["sum", "mean"]
})
```

---

## transform()

Devuelve un valor por cada fila original, en vez de resumir cada grupo en una sola fila

```python
df["promedio_sucursal"] = (
    df.groupby("sucursal")["ventas"].transform("mean")
)

df
```

```python
df["diferencia_promedio"] = (
    df["ventas"] - df["promedio_sucursal"]
)

df
```

---

## filter()

Evalua la condición sobre todos el grupo y conserva o elimina sus fila

```python
df.groupby("sucursal").filter(
    lambda grupo: grupo["ventas"].sum() > 40000
)
df
```

---

## Ordenar despues de agrupar

```python
ventas_producto = (
    df.groupby("producto", as_index=False)["ventas"]
    .sum()
    .sort_values("ventas", ascending=False)
)

ventas_producto
```
