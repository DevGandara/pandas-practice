# Transformaciones

Permiten modificar, convertir o crear información a partir de las columnas de un dataframe

axis=0 - Index o Renglones


axis=1 - Columnas

## Crear dataframe

```python
import pandas as pd
import numpy as np

data = {
    "producto": [" Laptop ", "mouse", "TECLADO", "Monitor", "Mouse", " Laptop "],
    "categoria": ["Tecnologia", "Accesorios", "Accesorios",
                  "Tecnologia", "Accesorios", "Tecnologia"],
    "precio": [15000, 300, 800, 4000, 300, 15500],
    "cantidad": [2, 5, 3, 1, 4, 1],
    "descuento": [0.10, 0, 0.05, 0.15, 0, 0.10],
    "fecha": ["2026-01-10", "2026-01-11", "2026-01-11",
              "2026-02-05", "2026-02-06", "2026-02-07"]
}

df = pd.DataFrame(data)
df
```

## Operaciones vectorizadas

```python
df["subtotal"] = df["precio"] * df["cantidad"]
df["precio_doble"] = df["precio"] * 2

df["total"] = df["subtotal"] * (1 - df["descuento"])

df[["producto", "precio", "cantidad", "descuento", "subtotal", "total"]]
```

---

## map()
Aplica una correspondencia o función a cada elemento de una Serie

```python
# Usando diccionario

categorias = {
    "Tecnologia": 1,
    "Accesorios": 2
}

df["categoria_codigo"] = df["categoria"].map(categorias)

df
```

```python
# Usa funcion lambda
df["precio_usd"] = df["precio"].map(lambda x: x / 18)

df
```

---

## apply()

Permite ejecutar funciones sobre una Serie o sobre las filas y columnas de un Dataframe

```python
def clasificar_precio(precio):
    if precio >= 5000:
        return "Alto"
    return "Bajo"

df["tipo_precio"] = df["precio"].apply(clasificar_precio)

df
```

```python
def clasificar_venta(fila):
    if fila["precio"] > 5000 and fila["cantidad"] >= 2:
        return "Importante"
    return "Normal"

df["tipo_venta"] = df.apply(clasificar_venta, axis=1)

df
```

---

## replace()

Permite sustituir valores especificos, manteniendo los demas sin modificaciones

```python
df["categoria_ingles"] = df["categoria"].replace({
    "Tecnologia": "Tech",
    "Accesorios": "Accessories"
})

df[["categoria", "categoria_ingles"]]
```

---

### Comparación entre map(), apply() y replace()

- map(): Transformar valores individualmente mediante un diccionario o función (Series)
- replace(): Sustituir valores especificos por otros (Series o DF)
- apply(): Ejecutar funciones personalizadas (Series o DF)


---

## np.where() y np.select()
Permite asignar valores dependiendo de una condición

```python
# Dos resultados posibles
df["venta_especial"] = np.where(
    (df["precio"] > 5000) & (df["cantidad"] >= 2),
    "Si",
    "No"
)

# Varias condiciones: se utiliza la primera que se cumpla
condiciones = [
    df["total"] >= 20000,
    df["total"] >= 5000
]

df["clasificacion"] = np.select(
    condiciones,
    ["Alta", "Media"],
    default="Baja"
)

df[["producto", "total", "venta_especial", "clasificacion"]]
```

## pd.cut()

Permite convertir valores numéricos en intervalos o grupos

```python
df["rango_precio"] = pd.cut(
    df["precio"],
    bins=[0, 1000, 5000, float("inf")], # 1 - 1000 Economico, 1001 - 5000 Medio, 5001+ Caro 
    labels=["Economico", "Medio", "Caro"]
)

df[["producto", "precio", "rango_precio"]]
```

## assign()

Crear columnas sin modificar directamente el DataFrame

```python
nuevo_df = df.assign(
    impuesto=lambda x: x["subtotal"] * 0.16,
    total_con_impuesto=lambda x: x["subtotal"] * 1.16
)
nuevo_df
```

---

## transform()

Transformar manteniendo las filas

```python
df["promedio_categoria"] = (
    df.groupby("categoria")["total"]
      .transform("mean")
)

df[["producto", "categoria", "total", "promedio_categoria"]]
```
