# Cargamos datos

```python
import pandas as pd

data = {
    'id_empleado': [101, 102, 103, 104, 105, 105, 106, 107, 108, 109, 110],
    'nombre': [' Ana López ', 'CARLOS GÓMEZ', 'maria rodrigues', 'Juan Pérez', 'Laura Martínez', 'Laura Martínez', '  pedro sánchez ', 'Elena Gomez', None, 'Luis Morales', 'SOFÍA SILVA'],
    'edad': ['28', '35', '42', 'DESCONOCIDO', '31', '31', '-5', '29', '45', 'NaN', '38'],
    'fecha_ingreso': ['2021-01-15', '15/02/2021', '2021/03/10', '2021-04-22', '2021-05-30', '2021-05-30', '2021.06.12', None, '2021-08-01', '2021-09-15', '2021-10-20'],
    'salario': ['$2,500.00', '$3,100.50', '2800', '$4,000.00', '$3,500.00', '$3,500.00', '$1,900.00', '9999999', '$3,200.00', None, '$4,100.00'],
    'departamento': [' Ventas ', 'ventas', '  IT', 'Finanzas', 'IT', 'IT', 'Recursos Humanos', 'ventas', 'Finanzas', '  IT ', 'RECURSOS HUMANOS'],
    'activo': ['Sí', 'SI', 'True', '1', 'No', 'No', '0', 'False', None, 'Sí', 'SI']
}

df = pd.DataFrame(data)

df
```

### Detectar valores faltantes

```python
df.isna() # Regresa valores True cuando hay celda vacia

df.isna().sum() # Suma de celdas vacias
```

### Eliminar valores vacios

```python
df = df.dropna()

df
```

### Contar duplicados

```python
df.duplicated()
```

### Eliminar duplicados

```python
df = df.drop_duplicates()

df
```

### Cambiar tipos de columnas

```python
df["id_empleado"] = df["id_empleado"].astype("string")

df["id_empleado"]
```

### Rellenar celdas vacias

```python
df["edad"] = pd.to_numeric(df["edad"], errors="coerce") # Convertir a none si no puede convertir a numeric

df.loc[df["edad"] <= 0, "edad"] = None # Cambiar a none edades negativas

df["edad"] = df["edad"].fillna(
    df["edad"].median()
)

df["edad"] = df["edad"].astype(int) # Regresamos a entero
df
```

## Renombrar columnas

```python
df = df.rename(columns={"salario":"salario esperado"})
df
```

## Limpiar textos

```python
df["nombre"] = df["nombre"].str.strip()
df["nombre"] = df["nombre"].str.lower()
df["nombre"] = df["nombre"].str.title()

df
```

## Reemplazar celdas

```python
mapeo = {
    "si": "Activo",
    "true": "Activo",
    "1": "Activo",
    "1.0": "Activo",  # Por si hay flotantes
    "no": "Inactivo",
    "false": "Inactivo",
    "0": "Inactivo",
    "0.0": "Inactivo",
}

df["activo"] = (
    df["activo"]
    .astype(str)
    .str.normalize("NFKD") # Normalizar para separar las palabras y los acentos b"S´i"
    .str.encode("ascii", errors="ignore") # Borra los caracteres especiales que pasa cuando se separan los acentos b"Si"
    .str.decode("utf-8") # Vuelve a transformar el texto en string "Si"
    .str.lower()
    .str.strip()
)

df["activo"] = df["activo"].replace(mapeo)
df
```
